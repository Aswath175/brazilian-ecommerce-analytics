**Missing Values and Duplicate Records Inspection**

 Identify missing values and duplicate records in all nine original Olist datasets. The code checks the total number of missing cells, the number of columns containing missing values, and the number of exact duplicate rows in each dataset. It also displays the data types and missing-value counts for individual columns. Finally, an overall data quality summary is generated to understand the condition of the original datasets before cleaning.

```python
#Inspect missing values and duplicate rows
PROJECT_DIR = Path("/content/drive/MyDrive/Olist_Capstone")
ORIGINAL_DIR = PROJECT_DIR / "01_Original_Data"

audit_results = []

for file in sorted(ORIGINAL_DIR.glob("*.csv")):
    df = pd.read_csv(file, low_memory=False)

    missing_cells = int(df.isna().sum().sum())
    columns_with_missing = int(df.isna().any().sum())
    duplicate_rows = int(df.duplicated().sum())

    audit_results.append({
        "Dataset": file.name,
        "Rows": len(df),
        "Missing Cells": missing_cells,
        "Columns With Missing Values": columns_with_missing,
        "Exact Duplicate Rows": duplicate_rows
    })

    print(f"\n{'=' * 55}")
    print("Dataset:", file.name)
    print("Data types:")
    print(df.dtypes.to_string())

    missing = df.isna().sum()
    missing = missing[missing > 0]

    print("\nColumns with missing values:")
    if missing.empty:
        print("None")
    else:
        print(missing.to_string())

audit_df = pd.DataFrame(audit_results)

print("\n\nOVERALL DATA QUALITY SUMMARY")
display(audit_df)
```

**Importing the Customers Dataset**

In this step, we set up the project folder paths and create the cleaned data folder if it does not already exist. Then, we load the original customers dataset using Pandas.

We perform initial data inspection to understand the dataset, including:


*   Checking the number of rows and columns.
*   Identifying exact duplicate rows.
*   Checking for missing and blank values.
*   Verifying the uniqueness of customer IDs and customer unique IDs.
*   Checking ZIP code ranges and customer state distribution.

This initial inspection helps us understand the quality of the original dataset before applying any cleaning operations.

```python
#importing customers dataset to cleaning
PROJECT_DIR = Path("/content/drive/MyDrive/Olist_Capstone")
ORIGINAL_DIR = PROJECT_DIR / "01_Original_Data"
CLEANED_DIR = PROJECT_DIR / "02_Cleaned_Data"

CLEANED_DIR.mkdir(parents=True, exist_ok=True)

# Read the original customers dataset
customers = pd.read_csv(
    ORIGINAL_DIR / "olist_customers_dataset.csv"
)

print("Original shape:", customers.shape)
print("Exact duplicate rows:", customers.duplicated().sum())

# Check missing values
print("\nMissing values:")
print(customers.isna().sum())

# Check primary and business key uniqueness
print("\nUnique customer_id values:",
      customers["customer_id"].nunique())

print("Unique customer_unique_id values:",
      customers["customer_unique_id"].nunique())

print("\nDuplicate customer_id values:",
      customers["customer_id"].duplicated().sum())

# Check ZIP code format and state values
print("\nCustomer ZIP code range:")
print(customers["customer_zip_code_prefix"].agg(["min", "max"]))

print("\nCustomer state counts:")
print(customers["customer_state"].value_counts().sort_index())

# Check blank text values
text_columns = ["customer_id", "customer_unique_id",
                "customer_city", "customer_state"]

for col in text_columns:
    blank_count = (
        customers[col].astype("string").str.strip().eq("").sum()
    )
    print(f"Blank values in {col}: {blank_count}")
```

**Checking Whitespace and ZIP Code Formatting**

In this step, we check the text columns for unnecessary spaces and verify the formatting of customer ZIP code prefixes.

The following checks are performed:

*   Identify text values containing extra spaces at the beginning or end.
*   Check whether customer ZIP code prefixes contain exactly 5 digits.
*   Count ZIP code prefixes that start with zero.

This step helps identify formatting inconsistencies and ensures that customer information is stored in a consistent format before further data cleaning.

```python
# Check whitespace and ZIP code formatting
for col in ["customer_id", "customer_unique_id",
            "customer_city", "customer_state"]:
    values = customers[col].astype("string")
    extra_spaces = values.ne(values.str.strip()).sum()
    print(f"{col} values with extra spaces: {extra_spaces}")

zip_codes = customers["customer_zip_code_prefix"].astype("string")

print("\nZIP prefixes not exactly 5 characters:",
      (~zip_codes.str.fullmatch(r"\d{5}", na=False)).sum())

print("ZIP prefixes starting with zero:",
      zip_codes.str.startswith("0", na=False).sum())
```

**Cleaning and Saving the Customers Dataset**

In this step, we create a cleaned copy of the original customers dataset and standardize the ZIP code prefixes.

The following cleaning operations are performed:

*   Load the original customers dataset using Pandas.
*   Read ZIP code prefixes as text to preserve leading zeros.
*   Standardize ZIP code prefixes to exactly five digits using zfill(5).
*   Validate the dataset by checking missing values, duplicate rows, and duplicate customer IDs.
*   Verify that all ZIP code prefixes contain exactly five digits.
*   Display sample ZIP code values to confirm the formatting.

After completing these checks, the cleaned dataset is saved as olist_customers_cleaned.csv in the 02_Cleaned_Data folder.

```python
PROJECT_DIR = Path("/content/drive/MyDrive/Olist_Capstone")
ORIGINAL_DIR = PROJECT_DIR / "01_Original_Data"
CLEANED_DIR = PROJECT_DIR / "02_Cleaned_Data"

CLEANED_DIR.mkdir(parents=True, exist_ok=True)

# Read ZIP prefixes as text to preserve leading zeros
customers_cleaned = pd.read_csv(
    ORIGINAL_DIR / "olist_customers_dataset.csv",
    dtype={"customer_zip_code_prefix": "string"}
)

# Standardize ZIP prefixes to five digits
customers_cleaned["customer_zip_code_prefix"] = (
    customers_cleaned["customer_zip_code_prefix"].str.zfill(5)
)

# Validate the result
print("Rows:", len(customers_cleaned))
print("Missing cells:", customers_cleaned.isna().sum().sum())
print("Exact duplicate rows:", customers_cleaned.duplicated().sum())
print("Duplicate customer IDs:",
      customers_cleaned["customer_id"].duplicated().sum())

print(
    "ZIP prefixes not exactly five digits:",
    (~customers_cleaned["customer_zip_code_prefix"]
     .str.fullmatch(r"\d{5}", na=False)).sum()
)

print("\nZIP prefix examples:")
print(customers_cleaned["customer_zip_code_prefix"].head(10).tolist())

# Save as a new file; keep the original unchanged
output_file = CLEANED_DIR / "olist_customers_cleaned.csv"
customers_cleaned.to_csv(output_file, index=False)

print("\nSaved file:", output_file)
print("File exists:", output_file.exists())
```

 **Cleaning and Saving the Order Items Dataset**

In this step, we prepare the order items dataset for further analysis by converting the shipping deadline column into the correct date and time format.

The following cleaning operations are performed:

*  Load the original order items dataset using Pandas.
*  Record the original number of rows before processing.
*  Convert the shipping_limit_date column into a datetime format using pd.to_datetime().
*  Use errors="coerce" to identify invalid date values by converting them into missing values.
*  Validate the dataset by checking missing cells and exact duplicate rows.
*  Check duplicate combinations of order_id and order_item_id.
*  Verify invalid shipping dates, non-positive prices, and negative freight values.
*  Save the prepared dataset as olist_order_items_cleaned.csv in the 02_Cleaned_Data folder.

```python
#cleaning and saving the order items dataset
PROJECT_DIR = Path("/content/drive/MyDrive/Olist_Capstone")
ORIGINAL_DIR = PROJECT_DIR / "01_Original_Data"
CLEANED_DIR = PROJECT_DIR / "02_Cleaned_Data"

CLEANED_DIR.mkdir(parents=True, exist_ok=True)

# Read the original dataset
items = pd.read_csv(
    ORIGINAL_DIR / "olist_order_items_dataset.csv",
    dtype={
        "order_id": "string",
        "product_id": "string",
        "seller_id": "string"
    }
)

original_rows = len(items)

# Convert shipping deadline to datetime
items["shipping_limit_date"] = pd.to_datetime(
    items["shipping_limit_date"],
    errors="coerce"
)

# Validate
print("Original rows:", original_rows)
print("Cleaned rows:", len(items))
print("Missing cells:", items.isna().sum().sum())
print("Exact duplicate rows:", items.duplicated().sum())

print(
    "Duplicate order-item combinations:",
    items.duplicated(
        subset=["order_id", "order_item_id"]
    ).sum()
)

print(
    "Invalid shipping dates:",
    items["shipping_limit_date"].isna().sum()
)

print("Non-positive prices:", (items["price"] <= 0).sum())
print("Negative freight values:", (items["freight_value"] < 0).sum())

# Save to the cleaned-data folder
output_file = CLEANED_DIR / "olist_order_items_cleaned.csv"
items.to_csv(output_file, index=False)

print("\nSaved file:", output_file)
print("File exists:", output_file.exists())
```

### Handling Missing Review Comments

In this step, we handle missing values in the customer review comment columns of the cleaned order reviews dataset.

- Load the previously saved cleaned reviews dataset.
- Record the original number of rows before making changes.
- Identify the two comment columns: `review_comment_title` and `review_comment_message`.
- Replace missing values in both columns with `No Comment`.
- Preserve all existing review records without deleting any rows.
- Save the updated dataset back to the cleaned-data folder.
- Validate the changes by checking the row count, remaining missing comment values, and exact duplicate rows.

**Validate all nine cleaned datasets together**

###  Validating All Nine Cleaned Datasets

In this step, we perform a final validation to compare the original datasets with their cleaned versions and ensure that all nine cleaned CSV files are saved correctly.

The following validation checks are performed:

- Define the original and cleaned filenames for all nine datasets.
- Check whether each original and cleaned file exists in the respective folders.
- Load each original and cleaned dataset using Pandas.
- Compare the row counts of the original and cleaned datasets.
- Identify whether any datasets have missing files or different row counts.
- Display a validation summary using a Pandas DataFrame.
- Calculate the total number of rows across all original datasets.
- Calculate the total number of rows across all cleaned datasets.

This validation helps confirm that all nine datasets are available and that no records have been unintentionally removed during the cleaning process.


```python
PROJECT_DIR = Path("/content/drive/MyDrive/Olist_Capstone")
ORIGINAL_DIR = PROJECT_DIR / "01_Original_Data"
CLEANED_DIR = PROJECT_DIR / "02_Cleaned_Data"

# Original filename : cleaned filename
file_pairs = {
    "olist_customers_dataset.csv":
        "olist_customers_cleaned.csv",

    "olist_geolocation_dataset.csv":
        "olist_geolocation_cleaned.csv",

    "olist_order_items_dataset.csv":
        "olist_order_items_cleaned.csv",

    "olist_order_payments_dataset.csv":
        "olist_order_payments_cleaned.csv",

    "olist_order_reviews_dataset.csv":
        "olist_order_reviews_cleaned.csv",

    "olist_orders_dataset.csv":
        "olist_orders_cleaned.csv",

    "olist_products_dataset.csv":
        "olist_products_cleaned.csv",

    "olist_sellers_dataset.csv":
        "olist_sellers_cleaned.csv",

    "product_category_name_translation.csv":
        "product_category_translation_cleaned.csv"
}

results = []

for original_name, cleaned_name in file_pairs.items():

    original_path = ORIGINAL_DIR / original_name
    cleaned_path = CLEANED_DIR / cleaned_name

    if not original_path.exists() or not cleaned_path.exists():
        results.append({
            "Dataset": original_name,
            "Original Rows": None,
            "Cleaned Rows": None,
            "Row Counts Match": False,
            "File Exists": cleaned_path.exists()
        })
        continue

    original_df = pd.read_csv(original_path)
    cleaned_df = pd.read_csv(cleaned_path)

    results.append({
        "Dataset": cleaned_name,
        "Original Rows": len(original_df),
        "Cleaned Rows": len(cleaned_df),
        "Row Counts Match": len(original_df) == len(cleaned_df),
        "File Exists": True
    })

validation = pd.DataFrame(results)

print(validation.to_string(index=False))

print("\nAll cleaned files exist:",
      validation["File Exists"].all())

print("All row counts match:",
      validation["Row Counts Match"].all())

print("\nTotal original rows:",
      validation["Original Rows"].sum())

print("Total cleaned rows:",
      validation["Cleaned Rows"].sum())
```
